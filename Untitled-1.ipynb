```python
import numpy as np
import matplotlib.pyplot as plt

def phase_portrait(A, title):
    x1 = np.linspace(-5, 5, 25)
    x2 = np.linspace(-5, 5, 25)
    X1, X2 = np.meshgrid(x1, x2)

    U = A[0,0]*X1 + A[0,1]*X2
    V = A[1,0]*X1 + A[1,1]*X2

    N = np.sqrt(U**2 + V**2)
    U, V = U/N, V/N

    plt.figure(figsize=(5,5))
    plt.quiver(X1, X2, U, V, color='teal', pivot='middle')
    plt.xlabel('$x_1$')
    plt.ylabel('$x_2$')
    plt.title(title)
    plt.grid(True)
    plt.axis('equal')
    plt.show()


A_list = {
    '(a)': np.array([[-8, -6], [0, -2]]),
    '(b)': np.array([[-8, -6], [0,  2]]),
    '(c)': np.array([[ 0,  4], [-4, 0]]),
    '(d)': np.array([[ 4, -4], [ 4, 4]]),
    '(e)': np.array([[ 1,  0], [ 0, 0]]),
    '(f)': np.array([[ 1,  0], [ 1, 0]]),
    '(g)': np.array([[ 0,  0], [ 0, 1]])
}

for k, A in A_list.items():
    phase_portrait(A, f'System {k}')
```

```python

x1 = np.linspace(-2*np.pi, 2*np.pi, 50)
x2 = np.linspace(-3, 3, 50)
X1, X2 = np.meshgrid(x1, x2)

U = X2
V = -np.sin(X1)

N = np.sqrt(U**2 + V**2)
U, V = U/N, V/N

plt.figure(figsize=(6,4))
plt.quiver(X1, X2, U, V, color='indianred')
plt.xlabel('$x$')
plt.ylabel('$\dot{x}$')
plt.title('Nonlinear System: $\ddot{x} + \sin(x) = 0$')
plt.grid(True)
plt.show()
```
